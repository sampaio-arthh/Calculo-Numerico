Primeiro, importaremos as bibliotecas com as quais trabalharemos.

```python
import seaborn
import numpy
```

Aqui anotaremos os valores importantes que serão usados para diferenciar as peças. Colocmaos como código mesmo que seja comentário, pois se juntar todas as instâncias de código, tem-se um código bem documentado e funcional.


```python
#Decorativa ou protótipo visual ~1%
#Tampa ou apoio s/ encaixe crítico ~0,5
#Espaçador ou componente de montagem ~0,1 a 0,2%
#Encaixe mecânico ou flange ~0,05 a 0,01%
#Vedação ou componente rotativo de precisão <0,05%
```

Agora, definiremos os valores no código e calcularemos as propriedades nominais do objeto:

```python
r = 50.00
h = 10.00
#mostrando as medidas salvas corretamente e exibindo truncadas para apenas as casas apresentadas como relevantes
print(f"{r:.2f}")
print(f"{h:.2f}")

delta_r = 0.05
delta_h = 0.02

#todas as medidas acima estão em mm

diametro = 2*r
perimetro = 2*numpy.pi*r
area = numpy.pi*r*r
volume = numpy.pi*r*r*h

grandezas = []
grandezas.append(numpy.round(diametro, 4))
grandezas.append(numpy.round(perimetro, 4))
grandezas.append(numpy.round(area,4))
grandezas.append(numpy.round(volume,4))


print(f"Diametro={numpy.round(diametro, 4):.4f}mm") #aqui usamos o format para 4 casas decimais para garantir que exiba as casas que o numpy.round não exibia
print(f"Perimetro={numpy.round(perimetro, 4)}mm")
print(f"Área={numpy.round(area, 4)}mm^2 | {numpy.round(area/10, 4)} cm^2")
print(f"Volume={numpy.round(volume, 4)}mm^3 (ou *10^-6 L)")
```

Em seguida calcularemos o erro relativo e erro percentual do raio e da espessura, para entender se as dimensões fornecidas são adequadas para os tipos de peça.

```python
erro_relativo_r = delta_r/r
erro_relativo_h = delta_h/h

erro_percentual_r = 100*erro_relativo_r
erro_percentual_h = 100*erro_relativo_h

print(f"E_r(r) = {erro_relativo_r}\nE_r(h) = {erro_relativo_h}\n")
print(f"E_%(r) = {erro_percentual_r}\nE_%(h) = {erro_percentual_h}")
```

Com esses resultados e com as tolerâncis fornecidas inicialmente, chegamos nas seguintes conclusões:

- O erro percentual apresentado pelo raio e espessura nos permite a produção de peças dos tipos: Decorativa/Protótipo visual, Tampa/Apoio sem encaixe crítico, Espaçador/Componente de montagem;
- As demais peças têm menor tolerância de erro, assim, não podem ser produzidas com os erros encontrados.



## Propagação de Erros
| Grandeza | Valor Nominal | Propagação Aproximada |
| -------- | ------------- | --------------------- |
| Diâmetro | D = 2r        | ΔD ≈ 2Δr e ΔD/D ≈ Δr/r|
| Perímetro| P=2πr         | ΔP/P ≈ Δr/r           |
| Área     | A=πr          | ΔA/A ≈ 2Δr/r          |
| Volume   |V = πr²        | ΔV/V ≈ 2Δr/r + Δh/h   |

Considerando essa tabela, calcularemos a propagação aproximada do erro em cada medida.

```python
propagacoes_aproximadas = {}
#colocaremos as propagações aproximadas num dicionário com key=grandeza e value=resultado da expressão

propagacoes_aproximadas.update({"Diametro": (erro_relativo_r)})
#é aproximademante o erro relativo de r, pois é uma multiplicação direta de uma constante com r, carregando o erro presente em r

propagacoes_aproximadas.update({"Perimetro": (erro_relativo_r)})
#é aproximademante o erro relativo de r, pois é uma multiplicação direta de duas constantes com r, carregando o erro presente em r

propagacoes_aproximadas.update({"Area": (2*erro_relativo_r)})
#! Achar explicação para esses dois exemplos !
propagacoes_aproximadas.update({"Volume": ((2*erro_relativo_r)+(erro_relativo_h)) })

for key, value in enumerate(propagacoes_aproximadas):
  print(f"{value} = {grandezas[key]} +- {propagacoes_aproximadas[value]}")
#sem medidas pois o erro relativo considera a escala do problema
```
