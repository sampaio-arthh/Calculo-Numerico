```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Sistema Ax = b
A = np.array([[10.0, -1.0, 2.0],
              [-1.0, 11.0, -1.0],
              [2.0, -1.0, 10.0]])

b = np.array([6.0, 22.0, -10.0])

x = np.zeros(3)  # x0 = (0, 0, 0)
tol = 1e-5
max_iter = 100

history = []
errors = []

# Iteracao k=0
history.append([0, x[0], x[1], x[2], "-"] )

for k in range(1, max_iter + 1):
    x_old = x.copy()

    # Gauss-Seidel
    x[0] = (6.0 + x_old[1] - 2.0 * x_old[2]) / 10.0
    x[1] = (22.0 + x[0] + x_old[2]) / 11.0
    x[2] = (-10.0 - 2.0 * x[0] + x[1]) / 10.0

    # Erro ||x(k) - x(k-1)||_inf
    error = np.max(np.abs(x - x_old))
    errors.append(error)

    history.append([k, x[0], x[1], x[2], f"{error:.6e}"])

    if error < tol:
        break

# Tabela
df = pd.DataFrame(history, columns=["k", "x(k)", "y(k)", "z(k)", "||x(k)-x(k-1)||_inf"])
print("--- TABELA DE ITERAÇÕES ---")
print(df.to_string(index=False))

print(f"\nNúmero de iterações realizadas: {k}")
print(f"Solução aproximada obtida: x = {x[0]:.6f}, y = {x[1]:.6f}, z = {x[2]:.6f}")

# Gráficos
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Gráfico a) Variáveis vs k
ks = range(k + 1)
x_vals = [h[1] for h in history]
y_vals = [h[2] for h in history]
z_vals = [h[3] for h in history]

ax1.plot(ks, x_vals, 'o-', label='x(k)')
ax1.plot(ks, y_vals, 's-', label='y(k)')
ax1.plot(ks, z_vals, '^-', label='z(k)')
ax1.set_xlabel('Iteração (k)')
ax1.set_ylabel('Valores das Incógnitas')
ax1.set_title('a) Evolução das Aproximações')
ax1.grid(True)
ax1.legend()

# Gráfico b) Erro vs k em Escala Logarítmica
ax2.semilogy(range(1, k + 1), errors, 'ro-')
ax2.set_xlabel('Iteração (k)')
ax2.set_ylabel('Erro ||x(k) - x(k-1)||_inf')
ax2.set_title('b) Erro entre Iterações (Escala Logarítmica)')
ax2.grid(True, which="both", ls="--")

plt.tight_layout()
plt.show()
```
